```python
import os, subprocess, sys
REPO = "https://github.com/SEU_USUARIO/xhakinha-clips-ai.git"   # <- troque pelo seu repositório (privado exige token nos Secrets)
if not os.path.isdir("/content/app"):
    subprocess.run(["git", "clone", "--depth", "1", REPO, "/content/app"], check=True)
else:
    subprocess.run(["git", "-C", "/content/app", "pull", "--ff-only"], check=False)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "/content/app/requirements-kaggle.txt"], check=True)
sys.path.insert(0, "/content/app")
```
